##### Analyzing Subsets of Survey Data

```python
said_no = df[df['BetterLife'] == 'No']
said_no.head(3)
```

```python
said_no.shape
```

```python
df['BetterLife'].value_counts()
```

```python
said_no['BetterLife'].value_counts()
```

```python
df['BetterLife'].value_counts(normalize=True)
```

```python
said_no = df[df['BetterLife'] == 'No']
said_yes = df[df['BetterLife'] == 'Yes']
```

df is telling Python to make said_no equivalent to the df DataFrame (our original data set), but then [df['BetterLife'] == 'No'] is telling Python to only include rows from df in which the answer in the 'BetterLife' column is equal to 'No'.

```python
print(said_no['Age'].mean(),
      said_yes['Age'].mean(),
      said_no['Age'].median(),
      said_yes['Age'].median()
     )
```

we can find out using the same Boolean trick we’ve been using, but instead of using == to check our conditional, we’ll use >= and <= since we want to filter for respondents whose 'Age' were 50 and up or 25 and down

```python
over50 = df[df['Age'] >= 50]
under25 = df[df['Age'] <= 25]
```

```python
print(over50['BetterLife'].value_counts(normalize=True))
print(under25['BetterLife'].value_counts(normalize=True))
```

```python
print(len(over50))
```

```python
print(len(under25))
```

#### Filtering More Specific Subsets

```python
# people who answered ‘Yes’ to the better life question and who live in India
```

```python
filtered_1 = df[(df['BetterLife'] == 'Yes') & (df['Country'] == 'India')]
```

```python
print(filtered_1['BetterLife'].value_counts())
```

```python
print(filtered_1['Country'].value_counts())
```

```python
# people Who Answered yes, over 50, Indian, do not code as a hobby, contribute to opensource occasionally
```

```python
filtered = df[(df['BetterLife'] == 'Yes') & (df['Age'] >= 50) & (df['Country'] == 'India') & ~(df['Hobbyist'] == 'Yes') & ~(df['OpenSourcer'] == 'Never')]
```
